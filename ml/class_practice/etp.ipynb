```python
X=df.drop("num",axis=1)
y=df['num']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3 ,random_state=42)

```

```python
from sklearn.naive_bayes import GaussianNB
model=GaussianNB()
model.fit(X_train,y_train)
pred=model.predict(X_test)

from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,pred))
```

```python
def predict_heart_disease(input_data):
    input_array = np.array(input_data).reshape(1, -1)
    prediction = model.predict(input_array)
    
    if prediction[0] == 1:
        return "Patient has Heart Disease"
    else:
        return "Patient does NOT have Heart Disease"
```

```python
sample_patient_disease = [
    64,   # age (older age → higher risk)
    1,    # Gender (1 = male)
    4,    # cp (asymptomatic chest pain)
    170,  # trestbps (high resting BP)
    418,  # chol (high cholesterol)
    0,    # fbs (fasting blood sugar)
    1,    # restecg (abnormal ECG)
    168,  # thalach (high max heart rate)
    1,    # exang (exercise induced angina)
    4.2,  # oldpeak (ST depression – strong indicator)
    2,    # slope (flat/down slope)
    3,    # ca (multiple vessels colored)
    10     # thal (reversible defect)
]


predict_heart_disease(sample_patient_disease)
```
