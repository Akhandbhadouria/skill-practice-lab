```python
import pandas as pd 
df=pd.read_csv("/Users/lambardaar/Documents/study_hub/python/ml/naive_bayes/titanic.csv")
df
```

```python
df.drop(['PassengerId','Name','SibSp','Parch','Ticket','Cabin','Embarked'],axis='columns',inplace=True)
df.head(4)

from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Sex']=le.fit_transform(df['Sex'])
df['Age'].fillna(df['Age'].mean(),inplace=True)
df['Fare'].fillna(df['Fare'].mean(), inplace=True)

 
```

```python
print(df.head())
target=df['Survived']
inputs=df.drop('Survived',axis=1)
```

```python
inputs.Age.isnull().value_counts()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(inputs, target, test_size=0.3 ,random_state=42)
X_train.isna().sum()
```

```python
from sklearn.naive_bayes import GaussianNB
model=GaussianNB()
model.fit(X_train,y_train)
pred=model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, pred)
```

```python
def pre(pclass, sex, age, fare):
    # Encode 'Sex' value
 
    # Prepare input DataFrame
    input_df = pd.DataFrame({
        'Pclass': [pclass],
        'Sex': [sex],
        'Age': [age],
        'Fare': [fare]
    })
    # Fit model on the full dataset
    X = df.drop('Survived', axis=1)
    y = df['Survived']
    model = GaussianNB()
    model.fit(X, y)
    # Predict for the input
    prediction = model.predict(input_df)
    return prediction[0]


```
