```python
import numpy as np
import pandas as pd 
import matplotlib as plt
# import  warnings 
# warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder,StandardScaler
```

```python
from sklearn.linear_model import Perceptron
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import Dropout
from tensorflow.keras.utils import to_categorical
from sklearn.metrics import classification_report,accuracy_score,confusion_matrix
```

```python
import seaborn as sns
df=sns.load_dataset('iris')
df
```

perceptron model


```python
X=df.drop('species',axis=1)
y=df['species']

le=LabelEncoder()
y_int=le.fit_transform(y)

X_train,X_test,y_train,y_test=train_test_split(X,y_int,test_size=.2,random_state=4,stratify=y_int)

sc=StandardScaler()
X_train_scaled=sc.fit_transform(X_train)
X_test_scaled=sc.transform(X_test)


pre=Perceptron(max_iter=1000,random_state=42)
pre.fit(X_train_scaled,y_train)

y_predection=pre.predict(X_test_scaled)

print(accuracy_score(y_test,y_predection))
```
