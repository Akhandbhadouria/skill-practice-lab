```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/Users/lambardaar/Downloads/skill-practice-lab/deep learning/dataset/Bank_Churn.csv")
df
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split


le=LabelEncoder()
df['Geography']=le.fit_transform(df['Geography'])
df['Gender']=le.fit_transform(df['Gender'])

df.drop(columns='Surname',inplace=True)

X=df.drop(columns="Exited")
y=df['Exited']

X_train, X_test, y_train, y_test = train_test_split(X, y)
sc = StandardScaler()
sc.fit(X_train)        # ✅ learn ONLY from training
X_train = sc.transform(X_train)

X_test = sc.transform(X_test)  # ✅ apply same scalingX_train.shape
```

```python
# %pip install tensorflow

from keras.layers import Dense #ensure that 👉 Current layer ka har neuron → previous layer ke har neuron se connected hota hai
from keras.models import Sequential #Sequential is a model container It lets you build a neural network layer by layer in a straight line (used for forward feed nn (INPUT -> OUTPUT))
from keras.layers import BatchNormalization


ann=Sequential( )

ann.add(Dense(6,input_dim=11,activation="relu")) #6 neurons ,input_dimensions =11 → 11 features ,activation = ReLU
ann.add(BatchNormalization())

ann.add(Dense(4,activation="relu"))
ann.add(BatchNormalization())

ann.add(Dense(2,activation="relu"))
ann.add(BatchNormalization())

ann.add(Dense(1,activation="sigmoid")) # 1 neuron → binary classification sigmoid → output between 0 and 1


ann.compile(optimizer='adam',loss="binary_crossentropy",metrics=['accuracy']) #Ye line model ko training ke liye ready karti hai. Iske bina model ko pata hi nahi hota ki: seekhna kaise hai galti measure kaise karni hai
# optimizer-> adam optimizer uses adaptive learning + momentum 
# for binary input we use binary_crossentropy because it punish confidence if the model is wrong and give reward if its correct 
# metrics=['accuracy'] → performance dikhane ke liye


ann.fit(X_train,y_train,batch_size=100,epochs=50) #👉 epochs = kitni baar model pura data dekh ke apne weights improve karega
```

```python
prd=ann.predict(X_test)
prd_data=[]
for i in prd:
    if i[0]<0.5:
        prd_data.append(0)
    else:
        prd_data.append(1)
print(prd_data)
```

```python
prd=ann.predict(X_train)
prd_data1=[]
for i in prd:
    if i[0]<0.5:
        prd_data1.append(0)
    else:
        prd_data1.append(1)
print(prd_data1)
```

```python
from sklearn.metrics import accuracy_score

ann.predict(X_train)
print(accuracy_score(y_test,prd_data)*100)
print(accuracy_score(y_train,prd_data1)*100)
```
