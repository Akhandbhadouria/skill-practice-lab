```python
# This cell imports all the necessary libraries for the project.
import numpy as np                      # Numerical operations, arrays, matrix computations
import pandas as pd                    # Data handling (DataFrames, CSVs, preprocessing)
import seaborn as sns                  # Statistical data visualization (heatmaps, plots)
import matplotlib.pyplot as plt        # Basic plotting (graphs, charts)
import warnings                        # Handles warning messages

warnings.filterwarnings('ignore')      # Suppresses unnecessary warnings

from sklearn.preprocessing import LabelEncoder, StandardScaler   # type: ignore
# LabelEncoder → converts categorical labels to numbers
# StandardScaler → normalizes data (mean=0, std=1)

from sklearn.model_selection import train_test_split  
# Splits dataset into training and testing sets

from sklearn.linear_model import Perceptron  
# Simple linear classifier (single-layer neural network)

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix  
# accuracy_score → calculates accuracy
# classification_report → precision, recall, F1-score
# confusion_matrix → shows correct vs incorrect predictions

from tensorflow.keras.models import Sequential  
# Builds neural network layer-by-layer (linear stack)

from tensorflow.keras.layers import Dense, Conv2D, Flatten, MaxPooling2D, Dropout  
# Dense → fully connected layer (used for classification)
# Conv2D → extracts features from images using filters
# Flatten → converts 2D feature maps into 1D vector
# MaxPooling2D → reduces size, keeps important features
# Dropout → prevents overfitting by randomly disabling neurons


from tensorflow.keras.utils import to_categorical
```

```python
# This cell loads the MNIST dataset, which is a large database of handwritten digits commonly used for training and testing image processing systems.
from tensorflow.keras.datasets import mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
# This cell normalizes the pixel values of the training and testing images by scaling them to a range of 0 to 1. This is a common preprocessing step that helps improve the performance of the neural network.
# getting the values in the range of 0-1.
x_test=x_train.astype("float32")/ 255.0
x_train=x_train.astype("float32")/ 255.0
```

```python
# This cell reshapes the flattened training and testing data back into 28x28 pixel images, which is the required input shape for the image-based models.
# converting a row to a 28x28
x_train_img=x_train.reshape(-1,28,28)
x_test_img=x_test.reshape(-1,28,28)
```

```python
# This cell defines, compiles, trains, and evaluates a simple Perceptron model.
# It first converts the integer labels to one-hot encoded vectors, then builds a single-layer neural network,
# trains it on the training data, and finally evaluates its accuracy on the test data.
y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)



perceptron = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(10, activation="softmax")
])

perceptron.compile(optimizer="sgd", loss="categorical_crossentropy", metrics=["accuracy"])

history_percp = perceptron.fit(x_train_img, y_train_cat, epochs=5, batch_size=32, validation_data=(x_test_img, y_test_cat), verbose=1)

acc_percp = perceptron.evaluate(x_test_img, y_test_cat, verbose=0)[1]
acc_percp
```

```python
# This cell builds a simple Artificial Neural Network (ANN) with two hidden layers.
# The model is compiled with the Adam optimizer and trained for 5 epochs.
# Finally, it's evaluated on the test set to measure its accuracy.
ann = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation="relu"),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])

ann.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])

history_ann = ann.fit(x_train_img, y_train_cat, epochs=5, batch_size=32, validation_data=(x_test_img, y_test_cat), verbose=1)

acc_ann = ann.evaluate(x_test_img, y_test_cat, verbose=0)[1]

acc_ann
```

```python
# This cell constructs, compiles, and trains a Convolutional Neural Network (CNN).
# The data is reshaped to include a channel dimension for the convolutional layers.
# The CNN architecture includes convolutional, pooling, and dropout layers to learn hierarchical features and prevent overfitting.
x_train_cnn = x_train.reshape(-1, 28, 28,1)
x_test_cnn = x_test.reshape(-1, 28, 28, 1)
cnn = Sequential([
    Conv2D(32, kernel_size=(3,3), activation="relu", input_shape=(28,28,1)),
    MaxPooling2D(pool_size=(2,2)),
    Conv2D(64, kernel_size=(3,3), activation="relu"),
    MaxPooling2D(pool_size=(2,2)),
    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.5),
    Dense(10, activation="softmax")
])

cnn.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])

history_cnn = cnn.fit(x_train_cnn, y_train_cat, epochs=5, batch_size=32, validation_data=(x_test_cnn, y_test_cat), verbose=1)
```

```python
# This cell evaluates the trained CNN model on the test dataset to determine its final accuracy.
acc_cnn = cnn.evaluate(x_test_cnn, y_test_cat, verbose=0)[1]
acc_cnn
```
