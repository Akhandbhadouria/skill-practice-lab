```python
                                                                #lowercasing text.
df['review']=df['review'].str.lower()
```

```python
# Apply the HTML tag removal and print the cleaned review at index 3
import re
def remove_htmltag(text):
    p=re.compile('<.*?>')
    return p.sub('',text)


print("before removal:->  ",df.iloc[3]['review'])
df['review'] = df['review'].apply(remove_htmltag)
print("after removal:->  ",df['review'].iloc[3])
```

```python
#Removing puntuation
str="hello there !"

import string
exclude=string.punctuation

def remove_string(text):
    return text.translate(str.maketrans(' ',' ',exclude))

new_str=remove_string(str)
print(new_str)
```

                                            REMOVING STOPWORDS

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

def remove_stopwords(text):
    stop_words = set(stopwords.words('english'))
    new_text = []

    for word in text.split():
        if word not in stop_words:
            new_text.append(word)

    return " ".join(new_text)

text = "this is a simple example to remove stopwords"
print(remove_stopwords(text))
```
