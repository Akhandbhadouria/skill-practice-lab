```python
import pandas as pd 
import numpy as np
data=[
    "my name is akhand",
    "the name is the akhand is the "
]

from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer()
BOW=cv.fit_transform(data)


print(cv.vocabulary_)
print(BOW[0].toarray())
print(BOW[1].toarray())
```
