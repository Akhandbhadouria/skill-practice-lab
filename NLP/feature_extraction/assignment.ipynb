```python
import pandas as pd
df=pd.read_csv('/Users/lambardaar/Downloads/skill-practice-lab/NLP/IMDB Dataset.csv')
df
```

```python
# text preprocessing
df['review']=df['review'].str.lower()

import re
def html_rem(data):
    p=re.compile('<.*?>')
    return p.sub('',data)

df['review']=df['review'].apply(html_rem)

# remove punctuation
import string
df['review'] = df['review'].str.replace(f"[{string.punctuation}]", "", regex=True)

#remove extra space
df['text'] = df['text'].str.strip()
```

```python
#corpus count
ans=0
for i in df['review']:
    ans+=len(i.split())
print(ans)

#vocalolary count
vocab = set(" ".join(df['review']).split())#make each row to a single string and then find length 
print(len(vocab))
```

```python
#bag of words
from sklearn.feature_extraction.text import CountVectorizer
import numpy as np
cv=CountVectorizer()
v=cv.fit_transform(df['review'])
print(v[0].toarray())
```
