```python
import gensim #Library used for topic modeling and word embeddings.
import os #Used for file and directory operations.

from nltk import sent_tokenize #Convert paragraph → list of sentences
from gensim.utils import simple_preprocess #It performs basic text preprocessing: lowercase conversion, punctuation removal,tokenization ,removes very short tokens

story=[]
data_dir = '/Users/lambardaar/Downloads/skill-practice-lab/NLP/datasets/got_data'
for filename in os.listdir(data_dir): #os.listdir() returns all filenames in the directory.
    file_path = os.path.join(data_dir, filename) #Safely combines directory + filename.: '/datasets/got_data/got1.txt'
    with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
        corpus = f.read()
        raw_sent = sent_tokenize(corpus)
        for sent in raw_sent:
            story.append(simple_preprocess(sent))


model = gensim.models.Word2Vec(
    window=10,
    min_count=2
)

model.build_vocab(story)
model.train(story,total_examples=model.corpus_count,epochs=model.epochs)

model.wv.most_similar('daenerys')
```
